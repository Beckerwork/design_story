**Beep:** „Die Roboter haben einen Datensatz mit der Körpergröße und Laufgeschwindigkeit von über 20 Kindern gespeichert. Jeder blaue Punkt ist ein Kind. Und sie nutzen diesen Datensatz, um für jede neue Person – also euch – die Geschwindigkeit zu schätzen."
