### 2 - Training des Modells mit Daten

<div style="display: flex; align-items: center; gap: 24px; margin: 12px 0;">
  <div style="flex: 2;">
    <p style="margin: 0 0 12px 0;"><strong>Beep:</strong> „In unserem Beispiel nehmen die Roboter an, dass die Geschwindigkeit mit der Körpergröße steigt und zwar linear. Linear bedeutet: Für jeden zusätzlichen Zentimeter Körpergröße ändert sich die Geschwindigkeit um einen festen Betrag. Ein linearer Zusammenhang kann sehr gut mit einer Geraden, die die allgemeine Form <code>Output = m * Input + c</code> hat, beschrieben werden. Jetzt werden die Werte für <code>m</code> und <code>c</code> so bestimmt, dass die Gerade möglichst nahe an den Datenpunkten liegt. Den Prozess, die besten Werte für <code>m</code> und <code>c</code> zu bestimmen, nennt man auch Training."</p>
    <p style="margin: 0 0 12px 0;"><strong>Ben:</strong> „Verstehe! Bestimmen die Roboter die Gerade auch durch ausprobieren?"</p>
    <p style="margin: 0;"><strong>Beep:</strong> „Bei einer Geraden gibt es zwei Möglichkeiten. Die erste Möglichkeit ist, die Werte für <code>m</code> und <code>c</code> exakt zu berechnen, sodass der Fehler möglichst gering ist. Dazu bestimmt man mathematisch die Werte für <code>m</code> und <code>c</code>, bei denen der Fehler minimal ist.<br><br>Die zweite Möglichkeit ist – wie du schon vermutet hast – ein schrittweises Ausprobieren. Dabei startet man mit zufälligen Werten für <code>m</code> und <code>c</code> und versucht dann mithilfe bestimmter Verfahren (beispielsweise dem Gradientenabstiegsverfahren), den Fehler systematisch zu verringern. Wird der Fehler von Versuch zu Versuch kleiner, ist das ein Zeichen dafür, dass der Prozess in die richtige Richtung läuft – also macht man weiter. Am Ende entsteht so ein Modell, das möglichst genau zu den vorhandenen Daten passt. Das Training endet, wenn die Gerade genau genug ist oder wenn eine bestimmte Anzahl an Versuchen durchgeführt wurde.<br><br>Du kannst auch noch einmal nach oben schauen: Die optimale Gerade der Roboter hatte die Werte <code>m = 0,33</code> und <code>c = -25,74</code>."</p>
